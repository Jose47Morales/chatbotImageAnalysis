```python
import os
import cv2
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Seleccionamos una imagen de muestra para el análisis
image_folder = '/content/images'
images = [f for f in os.listdir(image_folder) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

if not images:
    print(f"No se encontraron imágenes en {image_folder}")
else:
    sample_image_path = os.path.join(image_folder, images[8])
    print(f"Analizando la imagen: {sample_image_path}")

    # Cargar y preparar la imagen
    image = cv2.imread(sample_image_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image = cv2.resize(image, (150, 150))
    pixels = image.reshape(-1, 3)

    # Ajustar KMeans para k de 2 a 10 y registrar la inercia
    k_values = list(range(2, 11))
    inertias = []

    for k in k_values:
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
        kmeans.fit(pixels)
        inertias.append(kmeans.inertia_)

    # Guardar resultados en un DataFrame
    results_df = pd.DataFrame({
        'k': k_values,
        'inertia': inertias
    })

    display(results_df)

    # Graficar el método del codo (Elbow Method)
    plt.figure(figsize=(8, 5))
    plt.plot(k_values, inertias, marker='o', linestyle='-', color='b')
    plt.title('Método del Codo - Inercia vs. Número de Clústeres (k)')
    plt.xlabel('Número de Clústeres (k)')
    plt.ylabel('Inercia')
    plt.grid(True)
    plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import cv2

# --- PARTE 1: Proporción de píxeles por clúster con KMeans (k=6) original ---
k_6_original = KMeans(n_clusters=6, random_state=42, n_init=10)
labels_original = k_6_original.fit_predict(pixels)
centers_original = k_6_original.cluster_centers_.astype(int)

# Contar cuántos píxeles tiene cada clúster
pixel_counts = np.bincount(labels_original)
total_pixels = len(labels_original)
proportions = pixel_counts / total_pixels

original_proportions_df = pd.DataFrame({
    'Cluster Index': range(6),
    'RGB Color': [list(c) for c in centers_original],
    'Pixel Count': pixel_counts,
    'Proportion (%)': proportions * 100
}).sort_values(by='Proportion (%)', ascending=False)

print("=== PROPORCIÓN DE PÍXELES ORIGINAL (k=6) ===")
display(original_proportions_df)

# --- PARTE 2: Experimento - Excluir píxeles casi blancos (RGB > 235) ---
# Filtramos los píxeles donde R > 235, G > 235 y B > 235
white_threshold = 235
mask_not_white = ~((pixels[:, 0] > white_threshold) &
                   (pixels[:, 1] > white_threshold) &
                   (pixels[:, 2] > white_threshold))

filtered_pixels = pixels[mask_not_white]
print(f"\nFiltrado de fondo: Se removieron {total_pixels - len(filtered_pixels)} píxeles blancos. Quedan {len(filtered_pixels)} píxeles para el análisis.")

# Correr KMeans de 2 a 10 con los píxeles filtrados
k_values = list(range(2, 11))
filtered_inertias = []
filtered_silhouettes = []

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(filtered_pixels)
    filtered_inertias.append(kmeans.inertia_)

    # Silueta con muestra para optimizar velocidad
    score = silhouette_score(filtered_pixels, labels, sample_size=1000, random_state=42)
    filtered_silhouettes.append(score)

# --- PARTE 3: Graficar Codo y Silueta Filtrados ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 5))

# Gráfico del codo (Inercia)
ax1.plot(k_values, filtered_inertias, marker='o', color='b')
ax1.set_title('Método del Codo (Píxeles Filtrados)')
ax1.set_xlabel('Número de Clústeres (k)')
ax1.set_ylabel('Inercia')
ax1.grid(True)

# Gráfico de Silueta
ax2.plot(k_values, filtered_silhouettes, marker='o', color='g')
ax2.set_title('Coeficiente de Silueta (Píxeles Filtrados)')
ax2.set_xlabel('Número de Clústeres (k)')
ax2.set_ylabel('Silhouette Score')
ax2.grid(True)

plt.tight_layout()
plt.show()

# Mostrar tabla de siluetas
filtered_results_df = pd.DataFrame({
    'k': k_values,
    'silhouette_score': filtered_silhouettes
})
display(filtered_results_df)

# --- PARTE 4: Nueva Paleta para k=6 sin fondo blanco ---
k_6_filtered = KMeans(n_clusters=6, random_state=42, n_init=10)
k_6_filtered.fit(filtered_pixels)
centers_filtered = k_6_filtered.cluster_centers_.astype(int)

bar_filtered = np.zeros((60, 300, 3), dtype="uint8")
start_x = 0
for color in centers_filtered:
    end_x = start_x + (300 // 6)
    cv2.rectangle(bar_filtered, (start_x, 0), (end_x, 60), color.tolist(), -1)
    start_x = end_x

plt.figure(figsize=(10, 2))
plt.imshow(bar_filtered)
plt.title(f"Nueva Paleta k=6 (Píxeles Filtrados: {centers_filtered.tolist()})")
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# 1. Convertir la imagen original (redimensionada a 150x150) a HSV
image_hsv = cv2.cvtColor(image, cv2.COLOR_RGB2HSV)
pixels_hsv = image_hsv.reshape(-1, 3)

# Definir el rango para los tonos verde-azulados (Teal / Green-Blue) en OpenCV HSV:
# En OpenCV, H va de 0 a 180 (mapeando 0-360 grados).
# Los tonos verde-azulados/turquesas usualmente están entre 150° y 210° (en escala de 360°).
# Dividido por 2, esto corresponde aproximadamente a H entre 75 y 105.
lower_teal = np.array([75, 40, 40])
upper_teal = np.array([105, 255, 255])

# Crear máscara para los píxeles verde-azulados
teal_mask = cv2.inRange(image_hsv, lower_teal, upper_teal)
teal_pixel_count = np.sum(teal_mask > 0)
total_pixels_count = pixels_hsv.shape[0]
teal_proportion = (teal_pixel_count / total_pixels_count) * 100

print("=== ANÁLISIS DE TONOS VERDE-AZULADOS (TEAL) ===")
print(f"Píxeles identificados como verde-azulado: {teal_pixel_count} de {total_pixels_count}")
print(f"Proporción de color verde-azulado: {teal_proportion:.2f}%")

# 2. Filtrar el fondo usando HSV (Baja saturación y Alto brillo = Papel/Lienzo)
# S < 30 (baja saturación) y V > 200 (brillo alto)
low_sat_high_val_mask = (pixels_hsv[:, 1] < 30) & (pixels_hsv[:, 2] > 200)

# Filtrar los píxeles descartando este fondo
hsv_filtered_pixels_rgb = pixels[~low_sat_high_val_mask]

print(f"\n=== FILTRADO AVANZADO (HSV) ===")
print(f"Píxeles removidos (Fondo/Papel): {np.sum(low_sat_high_val_mask)} de {total_pixels_count}")
print(f"Píxeles restantes para clustering: {len(hsv_filtered_pixels_rgb)}")

# 3. Rerun KMeans (k=6) con los nuevos píxeles filtrados por HSV
k_6_hsv = KMeans(n_clusters=6, random_state=42, n_init=10)
k_6_hsv.fit(hsv_filtered_pixels_rgb)
centers_hsv_filtered = k_6_hsv.cluster_centers_.astype(int)

# Mostrar la nueva paleta resultante
bar_hsv_filtered = np.zeros((60, 300, 3), dtype="uint8")
start_x = 0
for color in centers_hsv_filtered:
    end_x = start_x + (300 // 6)
    cv2.rectangle(bar_hsv_filtered, (start_x, 0), (end_x, 60), color.tolist(), -1)
    start_x = end_x

plt.figure(figsize=(10, 2))
plt.imshow(bar_hsv_filtered)
plt.title(f"Paleta k=6 (Filtrado de Papel HSV): {centers_hsv_filtered.tolist()}")
plt.axis('off')
plt.show()
```

En una ilustración con gradientes y fondo de papel, filtrar el fondo con HSV mejora la paleta (los clusters dejan de gastarse en blancos), pero K-Means por área sigue ignorando los acentos minoritarios: los tonos verde-azulados son solo el 1,5 % de los píxeles y no obtienen un centro propio con k=6.

```python
import os
import cv2
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Seleccionamos una imagen de muestra para el análisis
image_folder = '/content/images'
images = [f for f in os.listdir(image_folder) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

if not images:
    print(f"No se encontraron imágenes en {image_folder}")
else:
    sample_image_path = os.path.join(image_folder, images[1])
    print(f"Analizando la imagen: {sample_image_path}")

    # Cargar y preparar la imagen según tu flujo:
    # 1. Leer imagen
    image = cv2.imread(sample_image_path)
    # 2. Convertir color a RGB
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    # 3. Redimensionar a 150x150
    image = cv2.resize(image, (150, 150))
    # 4. Aplanar los píxeles
    pixels = image.reshape(-1, 3)

    # Ajustar KMeans para k de 2 a 10 y registrar la inercia (inertia_)
    k_values = list(range(2, 11))
    inertias = []

    for k in k_values:
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
        kmeans.fit(pixels)
        inertias.append(kmeans.inertia_)

    # Guardar resultados en un DataFrame
    results_df = pd.DataFrame({
        'k': k_values,
        'inertia': inertias
    })

    display(results_df)

    # Graficar el método del codo (Elbow Method)
    plt.figure(figsize=(8, 5))
    plt.plot(k_values, inertias, marker='o', linestyle='-', color='b')
    plt.title('Método del Codo - Inercia vs. Número de Clústeres (k)')
    plt.xlabel('Número de Clústeres (k)')
    plt.ylabel('Inercia')
    plt.grid(True)
    plt.show()
```

```python
from sklearn.metrics import silhouette_score

if not images:
    print(f"No se encontraron imágenes en {image_folder}")
else:
    # Usamos la misma imagen y píxeles del paso anterior
    print(f"Calculando el coeficiente de silueta para la imagen: {sample_image_path}")

    silhouette_scores = []
    k_values = list(range(2, 11))

    for k in k_values:
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
        labels = kmeans.fit_predict(pixels)

        # Calculamos la silueta usando un subconjunto (sample_size) de los píxeles para acelerar la ejecución
        score = silhouette_score(pixels, labels, sample_size=1000, random_state=42)
        silhouette_scores.append(score)

    # Crear DataFrame con los nuevos resultados
    silhouette_df = pd.DataFrame({
        'k': k_values,
        'silhouette_score': silhouette_scores
    })

    display(silhouette_df)

    # Graficar el Coeficiente de Silueta por cada K
    plt.figure(figsize=(8, 5))
    plt.plot(k_values, silhouette_scores, marker='o', linestyle='-', color='g')
    plt.title('Coeficiente de Silueta vs. Número de Clústeres (k)')
    plt.xlabel('Número de Clústeres (k)')
    plt.ylabel('Silhouette Score')
    plt.grid(True)
    plt.show()
```

Con k=6, cinco clusters reparten los neutros en porciones similares (16-22 % cada uno) y solo un acento (azul, 4 %) obtiene cluster propio. Los demás colores llamativos de los productos no aparecen en la paleta.
